# full()

```python
arr1=np.full((3,4),'A')
print(arr1)
```

```python
arr2=np.full_like(arr1,5)
print(arr2)
print(arr2.dtype)
```

```python
arr3=np.full_like(arr2,66,int)
print(arr3)
print(arr3.dtype)
```

## 比较函数 any all

```python
arr1=np.array([1,2,3,4,5])
print(np.array(arr1>3))
```

```python
print(np.any(arr1>3))
```

```python
print(np.all(arr1>3))
```

## 排序函数

```python
print(arr1)
```

```python
arr1.sort(axis=0)
print(arr1)
```
