```python
df['sex']=df['sex'].astype('category')
```

```python
df.info()
```

```python
df['bill_ratio']=df['bill_length_mm']/df['bill_depth_mm']
df
```

```python
# 分箱 把体重分成三个等级
labels =['低','中','高']
df['mass_level']=pd.cut(df['body_mass_g'],bins=3,labels=labels)
```

```python
df
```

```python
df['mass_level'].value_counts()
```

```python
# 按岛屿，性别分组
df.groupby(['sex']).agg({
    'body_mass_g':['mean','count']
})
```

```python
# 按岛屿，性别分组
df.groupby(['island','sex']).agg({
    'body_mass_g':['mean','count']
})
```
