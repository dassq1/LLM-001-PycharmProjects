# 面向对象写法

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
%matplotlib inline
x=np.linspace(-10,10,100)
y1=np.sin(x)
y2=np.cos(x)
```

```python
fig,ax=plt.subplots(2,figsize=(10,6))
```

```python
ax[0].set_xlim(-10,10)
ax[0].set_ylim(-2,2)
ax[0].set_xlabel('x')
ax[0].set_ylabel('sin(x)')
ax[0].set_title('sin')
ax[0].plot(x,y1)

ax[1].plot(x,y2)
ax[1].set_xlabel('x')
ax[1].set_ylabel('cos(x)')
ax[1].set_title('cos')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 生成数据
x = np.linspace(-10, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

# 创建画布和子图
fig, ax = plt.subplots(2, figsize=(10, 6))

# 第一个子图（正弦）
ax[0].plot(x, y1, color='red')
ax[0].set_xlim(-10, 10)
ax[0].set_ylim(-2, 2)
ax[0].set_xlabel('x')
ax[0].set_ylabel('sin(x)')
ax[0].set_title('sin')
ax[0].grid(True)

# 第二个子图（余弦）
ax[1].plot(x, y2, color='blue')
ax[1].set_xlim(-10, 10)
ax[1].set_ylim(-2, 2)
ax[1].set_xlabel('x')
ax[1].set_ylabel('cos(x)')
ax[1].set_title('cos')
ax[1].grid(True)

# 调整间距并显示
plt.tight_layout()
plt.show()
```

## 多变量

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams["font.sans-serif"]=["SimHei"]
rcParams["axes.unicode_minus"]=False
```

```python
def year_color(x):
    match x.year:
        case 2012:
            return 'r'
        case 2013:
            return 'g'
        case 2014:
            return 'w'
        case 2015:
            return 'c'
df=pd.read_csv("data/weather.csv")
df['date']=pd.to_datetime(df['date'])
df['color']=df['date'].apply(year_color)
fig,ax1=plt.subplots()
ax1.scatter(df['temp_max'],df['precipitation'],c=df['color'])
ax1.set_xlabel("最高气温")
ax1.set_ylabel("降水量")
plt.show()
```

# Pandas可视化

```python
import pandas as pd
```

```python
df=pd.read_csv("data/sleep.csv")
df.info()
```

## 柱状图

```python
pd.cut(df['sleep_duration'],[0,5,6,7,8,9,10,11,12]).value_counts().plot.bar(color=['r','g','b','y','c','m','k','purple'])
```

## 折线图

```python
pd.cut(df['sleep_duration'],[0,5,6,7,8,9,10,11,12]).value_counts().sort_index().plot()
```

## 面积图

```python
pd.cut(df['sleep_duration'],[0,5,6,7,8,9,10,11,12]).value_counts().sort_index().plot.area
```
