1. 使用 Numpy 创建一个包含 1 到 10 的整数的一维数组，并将其重塑为 2 行 5 列的二维数组。
   提示：重塑方法arr.reshape(形状)

```python
import numpy as np
```

```python
arr1=np.array(range(1,11))
print(arr1)
print(arr1.reshape(2,5))
```

2.创建一个3*4的二维数组，并用[1,5]的随机整数进行填充

```python
arr1=np.random.randint(1,6,(3,4))
print(arr1)
```

3.   zeros()：返回给定形状和类型的新数组，用0填充。
     ones()：返回给定形状和类型的新数组，用1填充。
     empty()：返回给定形状和类型的未初始化的新数组。

```python
arr1=np.zeros((3,4),dtype=int)
print(arr1)
```

```python
arr2=np.ones((2,4),dtype=float)
print(arr2)
```

```python
arr3=np.empty((5,1),dtype=int)
print(arr3)
```

4.创建一个数组 统计 平均值 最值 方差  标准差

```python
arr1=np.array([[1,2,3,4],[3,4,5,6],[5,6,7,8]])
print(arr1)
```

```python
print(arr1.mean())
print(arr1.sum())
print(arr1.max())
print(arr1.min())
print(arr1.var())
print(arr1.std())
```

5.创建一个三行5列的数组进行排序
  指定不同的轴线 查看效果

```python
arr1=np.random.randint(0,10,(3,5))
print(arr1)
```

```python
print(np.sort(arr1))
```

```python
print(np.sort(arr1,axis=0))
```

```python
print(np.sort(arr1,axis=1))
```

```python
print(np.sort(arr1,axis=-1))
```

```python
import pandas as pd
s = pd.Series([11, 22, 22, 44])

print("线性插值 (linear):", s.quantile(0.25, interpolation='linear'))   # 默认值
print("较低值 (lower):"  , s.quantile(0.25, interpolation='lower'))
print("较高值 (higher):" , s.quantile(0.25, interpolation='higher'))
print("最近值 (nearest):", s.quantile(0.25, interpolation='nearest'))
print("中点值 (midpoint):", s.quantile(0.25, interpolation='midpoint'))
```
