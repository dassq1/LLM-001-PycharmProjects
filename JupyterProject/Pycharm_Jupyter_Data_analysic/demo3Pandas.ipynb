# DataFrame的布尔索引

```python
df = pd.DataFrame(
  data={"age": [20, 30, 40, 10], "name": ["张三", "李四", "王五", "赵六"]},
  columns=["name", "age"],
  index=[101, 104, 103, 102],
)
```

```python
print(df)
```

```python
print(df['age']>25)
```

```python
print(df[df['age']>25])
```
