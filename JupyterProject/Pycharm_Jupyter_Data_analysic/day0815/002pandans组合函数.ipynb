```python
df1=pd.DataFrame(data={'a':[1,2],'b':[4,5]},
                 index=[1,2])
df2=pd.Series(data=[7,10],index=[1,2],name='a')
```

```python
df2
```

```python
pd.concat([df1,df2])
```

```python
pd.concat([df1,df2],axis=1)
```

# 重置索引

```python
df1=pd.DataFrame(data={'a':[1,2],'b':[4,5]},index=[1,2])
df2=pd.DataFrame(data={'a':[7,8],'b':[10,11]},index=[1,2])
```

```python
pd.concat([df1,df2])
```

```python
pd.concat([df1,df2],ignore_index=True)
```

```python
df1=pd.DataFrame(data={'a':[1,2],'b':[4,5]},index=[1,2])
df2=pd.DataFrame(data={'b':[7,8],'c':[10,11]},index=[2,3])
```

```python
df1
```

```python
df2
```

```python
pd.concat([df1,df2])
```

```python
pd.concat([df1,df2],join='inner').loc[2]
```
