# 分词器

```python
text="!我是徐应lai"
tokens=tokenizer.tokenize(text)
print(tokens)
```

```python
ids=tokenizer.convert_tokens_to_ids(tokens)
print(ids)
```

```python
# encode 一步到位，默认加上 CLS  SEP
ids=tokenizer.encode(text,add_special_tokens=False)
print(ids)
tokens=tokenizer.convert_ids_to_tokens(ids)
print(tokens)
```

```python
# decode 一般bert不会用来生成句子，这个也都是继承自父类的方法
text=tokenizer.decode(ids,skip_special_tokens=True,)
print(text)
```

# 数据预处理

```python
# 删除列
train_dataset=train_dataset.remove_columns(['cat'])
print(train_dataset)
```

```python
# 过滤行
train_dataset.filter(lambda x:x['review'] is None)
```

```python
train_dataset=train_dataset.filter(lambda x:x['review'] is not None)
train_dataset
```

```python
# 划分数据集
dataset_dict=train_dataset.train_test_split(test_size=0.2,)
dataset_dict
```

```python
# map()函数，
# 逐条处理
from transformers import AutoTokenizer
tokenizer=AutoTokenizer.from_pretrained("google-bert/bert-base-chinese")
def map_func(examples):

    inputs =tokenizer( examples['review'],padding=False,truncation=True,)   # 先不pad,先不设置返回tesor，现在返回的还是默认的ID列表
    return inputs
maped_dataset_dict=dataset_dict.map(map_func,batched=False,remove_columns=['review'])
maped_dataset_dict=maped_dataset_dict.rename_column(original_column_name='label',new_column_name='labels')
print(maped_dataset_dict)
```

```python
# map()按批处理
def batch_map_func(batch):
    inputs=tokenizer( batch['review'],padding=False,truncation=True,)
    return inputs
mapped_dataset_dict=dataset_dict.map(batch_map_func,batched=True,remove_columns=['review'],batch_size=1000)
print(mapped_dataset_dict)
```

# 保存数据集

```python
mapped_dataset_dict.save_to_disk('./data/processed')
```

```python
from datasets import load_from_disk
load_dataset_dict=load_from_disk('./data/processed')
print(load_dataset_dict)
```
