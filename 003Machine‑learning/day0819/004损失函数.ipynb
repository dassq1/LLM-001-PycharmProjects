```python
from sklearn.metrics import mean_absolute_error,mean_squared_error
import numpy as np
```

```python
# 定义真实和预测值
y_true=[1,2,3,4]
y_pred=[6.2,2.2,3.3,4.5]
```

```python
loss_mae=mean_absolute_error(y_true,y_pred)
loss_mse=mean_squared_error(y_true,y_pred)
print(loss_mae)
print(loss_mse)# 对异常值更敏感
```

```python
# 二维
y_true=np.random.rand(10,3)
y_pred=np.random.rand(10,3)
# y_true
# y_pred
```

```python
print(mean_absolute_error(y_true, y_pred))
print(mean_squared_error(y_true, y_pred))
```
