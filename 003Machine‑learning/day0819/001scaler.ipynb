```python
import numpy as np
from sklearn.preprocessing import MinMaxScaler,StandardScaler
```

```python
X=np.array([[2,1],[3,1],[1,4],[2,6]])
print(X)
```

# 归一化

```python
scaler=MinMaxScaler()
X_scaled=scaler.fit_transform(X)
print(X_scaled)
```

```python
scaler=MinMaxScaler(feature_range=(-1,1))
X_scaled=scaler.fit_transform(X)
print(X_scaled)
```

# 标准化

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print(X_scaled)
```

```python
mean=np.mean(X,axis=0)
std=np.std(X,axis=0)
print((X-mean)/std)
```
