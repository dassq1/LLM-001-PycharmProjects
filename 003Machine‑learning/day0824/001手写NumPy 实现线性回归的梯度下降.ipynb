```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 准备数据
x=np.array([1,2,3,4,5],dtype=float)
y=np.array([3,5,7,9,11],dtype=float)
print(y.dtype)
```

```python
# 2.初始化模型参数
w=0.0
b=0.0
```

```python
# 3.设置超参数学习率
learning_rate=0.01
epochs=1000
```

```python
# 4.设置每次训练的损失
loss_history=[]
```

```python
# 5.开始梯度下降
for epoch in range(epochs):
    # 预测
    y_pred=w*x+b # x和y_pred都是向量
    # 计算误差
    error=y_pred-y
    # 计算损失
    loss=np.mean(error**2) #数组逐元素平方
    # print(loss)
    loss_history.append(loss)
    
    # 计算梯度
    dw=(2/len(x))*np.sum(error*x)
    db=(2/len(x))*np.sum(error)
    # 更新参数
    w= w-learning_rate*dw
    b=b-learning_rate*db
    
```

```python
# 6.输出结果
print('最终w=',w)
print('最终b=',b)
```

```python
# 7.绘制损失变化
plt.plot(loss_history[:20])
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Gradient Descent")
plt.show()
```
