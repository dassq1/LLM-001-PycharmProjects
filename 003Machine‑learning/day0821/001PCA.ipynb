```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
```

```python
n=1000
pc1=np.random.normal(0,1,n)
pc2=np.random.normal(0,0.5,n)
noise=np.random.normal(0,0.1,n)
```

```python
f1=pc1+pc2
f2=pc1-pc2
f3=pc1+noise
```

```python
print(f1.shape)
print(f2.shape)
print(f3.shape)
```

```python
x=np.vstack([f1,f2,f3]).T
print(x.shape)
# pca分析
pca=PCA(n_components=0.9)
x_pca=pca.fit_transform(x)
print(pca.explained_variance_ratio_)
print(x_pca.shape)
# 画图
# fig, axs = plt.subplots(1, 2, figsize=(12,4), subplot_kw={'projection': '3d'})
fig, axs = plt.subplots(1, 2, figsize=(12,4), projection='3d')
axs[0].scatter(x[:,0],x[:,1],x[:,2],c='g')
```
