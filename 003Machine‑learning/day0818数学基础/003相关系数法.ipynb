```python
import seaborn as sns
import pandas as pd
```

```python


advertising = pd.read_csv("data/advertising.csv")
advertising.head()
# advertising.describe()
# 预处理，去除ID列
advertising.drop(advertising.columns[0], axis=1, inplace=True)
advertising.head()
# 去除空值
advertising.dropna(inplace=True)
# 提取特征和标签
X = advertising.drop(columns='Sales')
Y = advertising['Sales']
print(X.shape)
print(Y.shape)
print(type(X))
print(type(Y))
print(Y.ndim) # 维度，Y必须是1维 ndim=1
# 计算皮尔逊相关系数
print("开始计算相关系数")
print(X.corrwith(Y, method='pearson'))
print("计算结束")
```
