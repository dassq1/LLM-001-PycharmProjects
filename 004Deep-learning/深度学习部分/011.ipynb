```python
import torch
import torch.nn as nn
```

```python
model=nn.Linear(1,1)
```

```python
# 数据
x = torch.tensor([[2.0]])
y=torch.tensor([[10.0]])
```

```python
# 损失函数
loss_fn=nn.MSELoss()
# 优化器
optimizer=torch.optim.SGD(
    model.parameters(),lr=0.1
)
```

```python
# 第一次训练
y_pred=model(x)
loss=loss_fn(y_pred,y)
loss.backward()
```

```python
print(model.weight.grad)
print(model.bias.grad)
print(loss)
```

```python
optimizer.step()
print(model.weight.grad)
print(model.bias.grad)
```

```python
optimizer.zero_grad()
print(model.weight.grad)
print(model.bias.grad)
```
