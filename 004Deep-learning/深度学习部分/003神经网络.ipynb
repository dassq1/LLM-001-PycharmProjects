```python
import torch
import torch.nn as nn
```

```python
# 定义全连接层
linear=nn.Linear(in_features=2,out_features=3,bias=True)
```

```python
print('weight',linear.weight)
print('weight',linear.weight.shape)
```

```python
print('bias:',linear.bias)
print('bias:',linear.bias.shape)
```

```python
# 前向传播
x=torch.tensor([[1,2],[3,4]]).float()
y=linear.forward(x)
```

```python
print(x.shape)
print(y.shape)
print(x)
print(y)
```

```python
y=linear(x)
y
```
