```python
import torch
```

```python
x=torch.tensor([100.0,3.0,10.0]) # x₁ = 100   房屋面积 房间 数量房龄
```

```python
# 定义权重
w=torch.tensor([2.0,5.0,-1.0])
```

```python
# 偏置
b=torch.tensor(10.0)
```

```python
# z
z=x@w+b
z.sum()
```

```python
import torch
import torch.nn as nn
# 一个线形层
layer=nn.Linear(3,2)
# 一个样本，3个特征
x=torch.tensor([[100.0,3.0,10.0]])
# 前向计算
y=layer(x)
print(y)
print(layer.modules)
```

```python
layer=nn.Linear(3,2)
x=torch.tensor([[1.0,2.0,3.0]])
y1=layer(x)
print(y1)
y2=x@layer.weight.T+layer.bias
print(y2)
```

```python
print(layer.weight)
print(layer.bias)
```
