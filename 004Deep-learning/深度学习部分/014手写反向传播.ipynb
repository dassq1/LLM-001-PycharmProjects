```python
import torch
```

```python
x=torch.tensor([[2.0,1.0]])
w1=torch.tensor([[0.5,0.2],
                 [0.3,0.8],
                 [-0.5,0.4]])
b1=torch.tensor([0.1,0.2,0.3])

w2=torch.tensor([[0.4,0.6,-0.2],
                 [-0.3,0.5,0.7]])
b2=torch.tensor([0.1,-0.1])
y=torch.tensor([0])
```

```python
z1=x@ w1.T +b1
print(z1)
```

```python
a1=torch.relu(z1)
```

```python
print(a1)
```

```python
z2=a1@ w2.T +b2
print(z2)
```

```python
p=torch.softmax(z2,dim=1)
print(p)
```

```python
loss1=-torch.log(p[0,y])
print(loss1)
loss2=-torch.log(p[0,1])
print(loss2)
```

```python
y_onehot=torch.nn.functional.one_hot(y,2)
print(y_onehot)
```

```python
dz2=p-y_onehot
print(dz2)
```

```python
dw2=dz2.T@a1
print(dw2)
db2=dz2.squeeze(0)
print(db2)
da1=dz2@ w2
print(da1)
```

```python
relu_grad=(z1>0).float()
print(relu_grad)
```

```python
dz1=da1*relu_grad
print(dz1)
```

```python
dw1=dz1.T@x
print(dw1)
```

```python
db1=dz1.squeeze(0)
print(db1)
```

```python
dx=dz1@w1
print(dx)
```

```python
lr = 0.1

w1 = w1 - lr * dw1
b1 = b1 - lr * db1

w2 = w2 - lr * dw2
b2 = b2 - lr * db2
print(w1)
```
