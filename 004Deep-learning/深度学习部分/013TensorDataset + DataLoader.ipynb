```python
import torch
from torch.utils.data import TensorDataset,DataLoader
```

```python
x=torch.tensor([[1],[2],[3],[4]])
y=torch.tensor([9,8,7,6])
ds=TensorDataset(x,y)
```

```python
print(ds[0])
print(ds[1])
```

```python
# DataLoader
train_dl=DataLoader(dataset=ds,batch_size=2,shuffle=True,drop_last=False)
print(x)
print(y)
```

```python
for bx ,by in train_dl:
    # print('bx shape:',bx.shape)
    # print('by shape:',by.shape)
    print(bx,by)

```
