```python
import torch
import torch.nn as nn

# 训练数据
x = torch.tensor([
    [1.0],
    [2.0],
    [3.0],
    [4.0]
])

y = torch.tensor([
    [3.0],
    [5.0],
    [7.0],
    [9.0]
])
```

```python
model=nn.Linear(1,1)
```

```python
loss_fn=nn.MSELoss()
```

```python
# 优化器
optimizer=torch.optim.SGD(
    model.parameters(),lr=0.01
)
```

```python
for epoch in range(100):
    y_pred=model(x)
    loss=loss_fn(y_pred,y)
    print(f"第{epoch}次的:",loss)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
```

```python
print("weight:", model.weight)
print("bias:", model.bias)
```
